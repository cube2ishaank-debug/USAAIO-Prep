```python
def dot_product(a,b):
  if len(a) != len(b):
    raise ValueError("Length of lists must be same") #Raise error if the length of lists are not same
  i=0
  dot=0
  while i < len(a): #iterate through each value in lists a & b
    dot+=a[i]*b[i] #add the product of corresponding number from a and b list to the answer
    i+=1 #move iteration to next number
  return dot #return dot product
```

## Exponents

While a normal math mode equation remains like text $x=2y-3$ while displayed math mode is displayed seperately like so: $$x=2y-3$$

To have multiple digits as exponents, there is a need for '{}' to be used. Without them an expression looks like $x^27$ and with them an expression looks like so $x^{27}$

Similarly, for having expressions as exponents, through '{}.'

Example: $x^{3y-7}$

## Subscript

Exponents act as superscripts, and to denote subscripts a '_' can be used.  

Example: $x_i$




Again, similar to superscript, to denote an expression as subscript, '{}' are required.

Example:
$x_{n-1}$

Nested subscripts can also be used, but need '{}' to denote to treat it's content as a seperate expression

For example, nested subscripts can look like so:

$x_{i_{n-3}}$

To show many items, a command called ldots, or lower dots can be utilized in LaTeX.

Example:
$a_1, a_2, a_3 \ldots a_{100}$

This is useful in the field of AI when denoting aspects such as weights for neural network perceptrons.

A perceptron:
$$p=\sigma(w_1i_1+b_1)$$
Here, $\sigma$ is the non-linear activation function, $w_1$ represents the weight for the first input, $i_1$ whose product is added to bias, $b_1$, before the activation function is applied.
