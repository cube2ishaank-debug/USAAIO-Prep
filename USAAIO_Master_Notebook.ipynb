```python
def dot_product(a,b):
  if len(a) != len(b):
    raise ValueError("Length of lists must be same") #Raise error if the length of lists are not same
  i=0
  dot=0
  while i < len(a): #iterate through each value in lists a & b
    dot+=a[i]*b[i] #add the product of corresponding number from a and b list to the answer
    i+=1 #move iteration to next number
  return dot #return dot product
```

Example: $x^{3y-7}$

Example: $x_i$

For example, nested subscripts can look like so:

### Introduction to Linear Algebra

#### Vectors & Matrices

Linear algebra is a branch of mathematics that works with vectors and matrixes.

A vector can be imagined as an arrow with a finite length. It is represented as an ordered list of numbers. An example of an array could be $\begin{bmatrix}1 &2 & 7 & -3 & 0 \end{bmatrix}$.

A matrix can be viewed as a list of vectors, where the matrix turns out to be rectangular, with columns representing multiple vectors. image.png




*Fun Fact: You can even think of a matrix as a vector of vectors!*

#### Properties of a Vetor & Matrix

##### Dimensions

A dimension of a vector, simply put, is the length of the vector. A dimension is important in ML, when any model expects an input of a certain dimension, passing an input with a different dimension often leads to the model to return an error.

 For example, $[1,-7,3,0,5.4]$ is a 5-dimensional vector.


Additionally, calculating the dimension of a matrix can be simply done by counting the height or the number of horizontal rows, $m$ and the width or the number of vertical columns, $n$. Finally, the dimension can be written as $m$x$n$.

For example, $ \begin{bmatrix}
1 & 0.43 & 3 \\
2 & 83 & 0
\end{bmatrix}$ has a dimension of 2x3.

##### Span

Each set of vectors, or a matrix has a span, which is the total space a set of vectors can cover when being scaled by any amount $k$ and being added to each other. For example, [1,0] and [0,1] have a span of $\mathbb{R}^{2}$. (NOTE: Total Space is denoted as $\mathbb{R}$)
